```python
import numpy as np
import torch
import physo
import physo.config.config0 as config0
from sklearn.metrics import r2_score, mean_squared_error
import os
import matplotlib.pyplot as plt

# Define the functions and their required operations
functions = {
    1: {'func': lambda x: x**3 + x**2 + x},
    2: {'func': lambda x: x**4 + x**3 + x**2 + x}, # assuming pow is available for n4
    3: {'func': lambda x: x**5 + x**4 + x**3 + x**2 + x}, # assuming pow is available for n4 and n5
    4: {'func': lambda x: x**6 + x**5 + x**4 + x**3 + x**2 + x},
    5: {'func': lambda x: np.sin(x**2)*np.cos(x) - 1},
    6: {'func': lambda x: np.sin(x) + np.sin(x + x**2)},
    7: {'func': lambda x: np.log(x + 1) + np.log(x**2 + 1)},
    8: {'func': lambda x: np.sqrt(x)},
    9: {'func': lambda x1, x2: np.sin(x1) + np.sin(x2**2)},
    10: {'func': lambda x1, x2: 2*np.sin(x1)*np.cos(x2)},
    11: {'func': lambda x1, x2: np.power(x1, x2)},
    12: {'func': lambda x, y_in: x**4 - x**3 + 0.5 * y_in**2 - y_in},
} 


# Dictionary to store results for plotting
plotting_data = {}

# Iterate through each function
for func_id, func_info in functions.items():
    f = func_info['func']
    op_names = ["cos","sin","add", "sub", "mul", "div","sqrt","log"] 

    print(f"\n--- Running for Function {func_id} ---")

    # --- 1. Data Preparation ---
    if func_id in [1, 2, 3, 4, 5, 6,7,8]:
        dim = 1
        x_original = np.linspace(-1, 1, 100)
        X_train = torch.tensor(x_original, dtype=torch.float32).view(1, -1)
        y_train = torch.tensor(f(x_original), dtype=torch.float32)

        x_val = np.linspace(-2, 2, 50) # Validation data
        X_val = torch.tensor(x_val, dtype=torch.float32).view(1, -1)
        y_val = torch.tensor(f(x_val), dtype=torch.float32)

        X_names = ["x"]

    elif func_id in [9, 10, 11, 12]:
        dim = 2
        x1_original = np.linspace(-1, 1, 10)
        x2_original = np.linspace(-1, 1, 10)
        x1_mesh, x2_mesh = np.meshgrid(x1_original, x2_original)
        X_train = torch.tensor(np.array([x1_mesh.flatten(), x2_mesh.flatten()]), dtype=torch.float32)
        y_train = torch.tensor(f(x1_mesh.flatten(), x2_mesh.flatten()), dtype=torch.float32)

        x1_val = np.linspace(-2, 2, 5)
        x2_val = np.linspace(-2, 2, 5)
        x1_val_mesh, x2_val_mesh = np.meshgrid(x1_val, x2_val)
        X_val = torch.tensor(np.array([x1_val_mesh.flatten(), x2_val_mesh.flatten()]), dtype=torch.float32)
        y_val = torch.tensor(f(x1_val_mesh.flatten(), x2_val_mesh.flatten()), dtype=torch.float32)

        X_names = ["x1", "x2"]


    # --- 2. Run physo Algorithm ---
    expression, logs = physo.SR(X_train, y_train,
        X_names         = X_names,
        y_name          = "y",
        op_names        = op_names,
        run_config      = config0.config0,
        epochs          = 30,
    )

    # --- 3. Evaluate Results ---
    if expression is not None:
        try:
            # Access the best expression directly from the returned object
            best_expr = expression

            # Calculate R^2 and NRMSE for training data
            y_train_pred = best_expr(X_train).detach().numpy()
            r2_train = r2_score(y_train.numpy(), y_train_pred)
            y_train_range = y_train.numpy().max() - y_train.numpy().min()
            nrmse_train = np.sqrt(mean_squared_error(y_train.numpy(), y_train_pred)) / y_train_range if y_train_range != 0 else float('inf')


            # Calculate R^2 and NRMSE for validation data
            y_val_pred = best_expr(X_val).detach().numpy()
            r2_val = r2_score(y_val.numpy(), y_val_pred)
            y_val_range = y_val.numpy().max() - y_val.numpy().min()
            nrmse_val = np.sqrt(mean_squared_error(y_val.numpy(), y_val_pred)) / y_val_range if y_val_range != 0 else float('inf')


            # --- 4. Display Results ---
            print(f"Function {func_id}:")
            print(f"  Discovered Expression: {best_expr.get_infix_sympy()}")
            print(f"  Training R^2: {r2_train:.4f}, NRMSE: {nrmse_train:.4f}")
            print(f"  Validation R^2: {r2_val:.4f}, NRMSE: {nrmse_val:.4f}")

            # Store data for plotting
            plotting_data[func_id] = {
                'X_val': X_val,
                'y_val': y_val,
                'best_expr': best_expr,
                'X_names': X_names,
                'y_name': "y",
                'func': f, # Store the original function
                'dim': dim # Store the dimension
            }


        except Exception as e:
            print(f"Error processing results for Function {func_id}: {e}")

    else:
        print(f"Function {func_id}: No expression found.")
```

```python
def plot_nguyen_fit(func_id, model, metrics=None, x_range=(-2, 2), n_points=400):
    """
    Grafica la función real del benchmark Nguyen y la predicha por el modelo EQL.
    Muestra además la ecuación recuperada y el R² en el título.
    """
    # Determinar dimensionalidad
    if func_id in [1, 2, 3, 4, 5, 6, 7, 8]:
        dim = 1
    elif func_id in [9, 10, 11, 12]:
        dim = 2
    else:
        raise ValueError("func_id debe ser 1–12.")


   # 1D: curva real vs predicha
    if dim == 1:
        x = np.linspace(x_range[0], x_range[1], n_points).astype(np.float32)
        if func_id == 1:
            f_true = x**3 + x**2 + x
        elif func_id == 2:
            f_true = x**4 + x**3 + x**2 + x
        elif func_id == 3:
            f_true = x**5 + x**4 + x**3 + x**2 + x
        elif func_id == 4:
            f_true = np.sin(x**2)
        elif func_id == 5:
            f_true = np.sin(x**2)*np.cos(x) - 1
        elif func_id == 6:
            f_true = np.sin(x) + np.sin(x + x**2)
        elif func_id == 7:
            f_true = np.log(x + 1) + np.log(x**2 + 1)
        elif func_id == 8:
            f_true = np.sqrt(x)

        X = x.reshape(-1, 1)
        f_pred = model.predict(X, verbose=0).flatten()

        r2_val = metrics["R2_test"] if metrics else r2_score(f_true, f_pred)

        plt.figure(figsize=(8,4))
        plt.plot(x, f_true, label="Función real", linewidth=2, color='C0')
        plt.plot(x, f_pred, '--', label="Predicha (EQL)", linewidth=2, color='C1')
        plt.xlabel("x", fontsize=12)
        plt.ylabel("y", fontsize=12)
        plt.title(f"Nguyen-{func_id} | R²_test = {r2_val:.3f}",
                  fontsize=14, weight='bold')
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()

    # 2D: mapa de contornos
    elif dim == 2:
        n_points_2d = 200
        x1 = np.linspace(x_range[0], x_range[1], n_points_2d)
        x2 = np.linspace(x_range[0], x_range[1], n_points_2d)
        X1, X2 = np.meshgrid(x1, x2)

        if func_id == 9:
            f_true = np.sin(X1) + np.sin(X2**2)
        elif func_id == 10:
            f_true = 2*np.sin(X1)*np.cos(X2)

        X_flat = np.stack([X1.flatten(), X2.flatten()], axis=1).astype(np.float32)
        f_pred = model.predict(X_flat, verbose=0).reshape(X1.shape)

        # R² en test si existe
        r2_val = metrics["R2_test"] if metrics else None

        # --- figura ---
        fig, axs = plt.subplots(1, 2, figsize=(10,4))
        cs0 = axs[0].contourf(X1, X2, f_true, levels=50, cmap='viridis')
        cs1 = axs[1].contourf(X1, X2, f_pred, levels=50, cmap='viridis')
        fig.colorbar(cs0, ax=axs[0]); fig.colorbar(cs1, ax=axs[1])
        axs[0].set_title("Función real", weight='bold')
        axs[1].set_title(f"Predicha (EQL)" + (f"\nR²_test = {r2_val:.3f}" if r2_val else ""), weight='bold')
        for ax in axs:
            ax.set_xlabel("x₁"); ax.set_ylabel("x₂")
        plt.tight_layout()
        plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Assuming plotting_data dictionary is available in the environment

if 'plotting_data' in locals() and plotting_data:
    for func_id in [9, 10, 11, 12]:
        if func_id in plotting_data and plotting_data[func_id]['X_val'].shape[0] == 2:
            data = plotting_data[func_id]
            X_val = data['X_val']
            y_val = data['y_val']
            best_expr = data['best_expr']
            X_names = data['X_names']
            y_name = data['y_name']

            # Generate a finer grid for contour plotting
            n_points_2d = 50 # You can adjust this for smoother contours
            x1_min, x1_max = X_val[0, :].min().item(), X_val[0, :].max().item()
            x2_min, x2_max = X_val[1, :].min().item(), X_val[1, :].max().item()

            x1_plot = np.linspace(x1_min, x1_max, n_points_2d)
            x2_plot = np.linspace(x2_min, x2_max, n_points_2d)
            X1_plot, X2_plot = np.meshgrid(x1_plot, x2_plot)

            # Prepare data for prediction on the new grid
            X_plot_flat = torch.tensor(np.array([X1_plot.flatten(), X2_plot.flatten()]), dtype=torch.float32)

            # Get predicted values on the new grid
            y_val_pred_plot = best_expr(X_plot_flat).detach().numpy().reshape(X1_plot.shape)

            # Get real values on the new grid (re-calculate based on the original function)
            # This assumes you have access to the original function definition
            # If not, you would need to interpolate from the original y_val data
            # For this example, we will use the original function definitions
            if func_id == 9:
                y_val_real_plot = (np.sin(X1_plot) + np.sin(X2_plot**2))
            elif func_id == 10:
                y_val_real_plot = (2*np.sin(X1_plot)*np.cos(X2_plot))
            elif func_id == 11:
                y_val_real_plot = (np.power(X1_plot, X2_plot))
            elif func_id == 12:
                y_val_real_plot = (X1_plot**4 - X1_plot**3 + 0.5 * X2_plot**2 - X2_plot)

            # --- Plotting ---
            # Create a figure for Real Data
            fig_real, ax_real = plt.subplots(figsize=(7, 6)) # Adjusted figure size for single plot
            cs0 = ax_real.contourf(X1_plot, X2_plot, y_val_real_plot, levels=50, cmap='viridis') # Using 'viridis'
            fig_real.colorbar(cs0, ax=ax_real, label='Real Data Value')
            ax_real.set_title(f'Function {func_id}: Real Data')
            ax_real.set_xlabel(X_names[0], fontsize=10)
            ax_real.set_ylabel(X_names[1], fontsize=10)
            ax_real.tick_params(axis='both', which='major', labelsize=8)
            plt.tight_layout()
            plt.show()

            # Create a figure for Predicted Data
            fig_pred, ax_pred = plt.subplots(figsize=(7, 6)) # Adjusted figure size for single plot
            cs1 = ax_pred.contourf(X1_plot, X2_plot, y_val_pred_plot, levels=50, cmap='viridis') # Using 'viridis'
            fig_pred.colorbar(cs1, ax=ax_pred, label='Predicted Data Value')
            ax_pred.set_title(f'Function {func_id}: Predicted Data')
            ax_pred.set_xlabel(X_names[0], fontsize=10)
            ax_pred.set_ylabel(X_names[1], fontsize=10)
            ax_pred.tick_params(axis='both', which='major', labelsize=8)
            plt.tight_layout()
            plt.show()


        elif func_id in plotting_data and plotting_data[func_id]['X_val'].shape[0] != 2:
             print(f"Plotting skipped for Function {func_id} (not 2-dimensional).")
        else:
             print(f"Data for Function {func_id} not available for plotting.")

else:
    print("No data available for plotting.")
```
