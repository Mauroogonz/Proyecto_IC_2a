```python
import numpy as np
import torch
import physo
import physo.config.config0 as config0
from sklearn.metrics import r2_score, mean_squared_error
import os
import matplotlib.pyplot as plt
```

```python


# Define the functions and their required operations
functions = {
    1: {'func': lambda x: x**3 + x**2 + x},
    2: {'func': lambda x: x**4 + x**3 + x**2 + x}, 
    3: {'func': lambda x: x**5 + x**4 + x**3 + x**2 + x}, 
    4: {'func': lambda x: x**6 + x**5 + x**4 + x**3 + x**2 + x},
    5: {'func': lambda x: np.sin(x**2)*np.cos(x) - 1},
    6: {'func': lambda x: np.sin(x) + np.sin(x + x**2)},
    7: {'func': lambda x: np.log(x + 1) + np.log(x**2 + 1)},
    8: {'func': lambda x: np.sqrt(x)},
    9: {'func': lambda x1, x2: np.sin(x1) + np.sin(x2**2)},
    10: {'func': lambda x1, x2: 2*np.sin(x1)*np.cos(x2)},
    11: {'func': lambda x1, x2: x1**x2},
    12: {'func': lambda x1, x2: x1**4 - x1**3 + 0.5 * x2**2 - x2},
} 


# Dictionary to store results for plotting
plotting_data = {}

# Iterate through each function
for func_id, func_info in functions.items():
    f = func_info['func']
    op_names = ["cos","sin","add", "sub", "mul", "div","sqrt","log"] 

    print(f"\n--- Running for Function {func_id} ---")

    # --- 1. Data Preparation ---
    if func_id in [1, 2, 3, 4, 5, 6,]:
        dim = 1
        x_original = np.linspace(-1, 1, 100)
        X_train = torch.tensor(x_original, dtype=torch.float32).view(1, -1)
        y_train = torch.tensor(f(x_original), dtype=torch.float32)

        x_val = np.linspace(-2, 2, 50) # Validation data
        X_val = torch.tensor(x_val, dtype=torch.float32).view(1, -1)
        y_val = torch.tensor(f(x_val), dtype=torch.float32)

        X_names = ["x"]
    elif func_id in [7]:
        dim = 1
        x_original = np.linspace(0, 2, 100)
        X_train = torch.tensor(x_original, dtype=torch.float32).view(1, -1)
        y_train = torch.tensor(f(x_original), dtype=torch.float32)

        x_val = np.linspace(0, 4, 50) # Validation data
        X_val = torch.tensor(x_val, dtype=torch.float32).view(1, -1)
        y_val = torch.tensor(f(x_val), dtype=torch.float32)

        X_names = ["x"]
    
    elif func_id in [8]:
        dim = 1
        x_original = np.linspace(0, 4, 100)
        X_train = torch.tensor(x_original, dtype=torch.float32).view(1, -1)
        y_train = torch.tensor(f(x_original), dtype=torch.float32)

        x_val = np.linspace(0, 6, 50) # Validation data
        X_val = torch.tensor(x_val, dtype=torch.float32).view(1, -1)
        y_val = torch.tensor(f(x_val), dtype=torch.float32)

        X_names = ["x"]

    elif func_id in [9, 10, 11, 12]:
        dim = 2
        x1_original = np.linspace(-1, 1, 10)
        x2_original = np.linspace(-1, 1, 10)
        x1_mesh, x2_mesh = np.meshgrid(x1_original, x2_original)
        X_train = torch.tensor(np.array([x1_mesh.flatten(), x2_mesh.flatten()]), dtype=torch.float32)
        y_train = torch.tensor(f(x1_mesh.flatten(), x2_mesh.flatten()), dtype=torch.float32)

        x1_val = np.linspace(-2, 2, 5)
        x2_val = np.linspace(-2, 2, 5)
        x1_val_mesh, x2_val_mesh = np.meshgrid(x1_val, x2_val)
        X_val = torch.tensor(np.array([x1_val_mesh.flatten(), x2_val_mesh.flatten()]), dtype=torch.float32)
        y_val = torch.tensor(f(x1_val_mesh.flatten(), x2_val_mesh.flatten()), dtype=torch.float32)

        X_names = ["x1", "x2"]


    # --- 2. Run physo Algorithm ---
    expression, logs = physo.SR(X_train, y_train,
        X_names         = X_names,
        y_name          = "y",
        op_names        = op_names,
        run_config      = config0.config0,
        epochs          = 30,
    )

    # --- 3. Evaluate Results ---
    if expression is not None:
        try:
            # Access the best expression directly from the returned object
            best_expr = expression

            # Calculate R^2 and NRMSE for training data
            y_train_pred = best_expr(X_train).detach().numpy()
            r2_train = r2_score(y_train.numpy(), y_train_pred)
            y_train_range = y_train.numpy().max() - y_train.numpy().min()
            nrmse_train = np.sqrt(mean_squared_error(y_train.numpy(), y_train_pred)) / y_train_range if y_train_range != 0 else float('inf')


            # Calculate R^2 and NRMSE for validation data
            y_val_pred = best_expr(X_val).detach().numpy()
            r2_val = r2_score(y_val.numpy(), y_val_pred)
            y_val_range = y_val.numpy().max() - y_val.numpy().min()
            nrmse_val = np.sqrt(mean_squared_error(y_val.numpy(), y_val_pred)) / y_val_range if y_val_range != 0 else float('inf')


            # --- 4. Display Results ---
            print(f"Function {func_id}:")
            print(f"  Discovered Expression: {best_expr.get_infix_sympy()}")
            print(f"  Training R^2: {r2_train:.4f}, NRMSE: {nrmse_train:.4f}")
            print(f"  Validation R^2: {r2_val:.4f}, NRMSE: {nrmse_val:.4f}")

            # Store data for plotting
            plotting_data[func_id] = {
                'X_val': X_val,
                'y_val': y_val,
                'best_expr': best_expr,
                'X_names': X_names,
                'y_name': "y",
                'func': f, # Store the original function
                'dim': dim # Store the dimension
            }


        except Exception as e:
            print(f"Error processing results for Function {func_id}: {e}")

    else:
        print(f"Function {func_id}: No expression found.")
```

```python
def plot_nguyen_fit(func_id, model, metrics=None, x_range=(-2, 2), n_points=400):
    """
    Grafica la función real del benchmark Nguyen y la predicha por el modelo EQL.
    Muestra además la ecuación recuperada y el R² en el título.
    """
    # Determinar dimensionalidad
    if func_id in [1, 2, 3, 4, 5, 6, 7, 8]:
        dim = 1
    elif func_id in [9, 10, 11, 12]:
        dim = 2
    else:
        raise ValueError("func_id debe ser 1–12.")


   # 1D: curva real vs predicha
    if dim == 1:
        x = np.linspace(x_range[0], x_range[1], n_points).astype(np.float32)
        if func_id == 1:
            f_true = x**3 + x**2 + x
        elif func_id == 2:
            f_true = x**4 + x**3 + x**2 + x
        elif func_id == 3:
            f_true = x**5 + x**4 + x**3 + x**2 + x
        elif func_id == 4:
            f_true = np.sin(x**2)
        elif func_id == 5:
            f_true = np.sin(x**2)*np.cos(x) - 1
        elif func_id == 6:
            f_true = np.sin(x) + np.sin(x + x**2)
        elif func_id == 7:
            f_true = np.log(x + 1) + np.log(x**2 + 1)
        elif func_id == 8:
            f_true = np.sqrt(x)

        X = x.reshape(-1, 1)
        f_pred = model.predict(X, verbose=0).flatten()

        r2_val = metrics["R2_test"] if metrics else r2_score(f_true, f_pred)

        plt.figure(figsize=(8,4))
        plt.plot(x, f_true, label="Función real", linewidth=2, color='C0')
        plt.plot(x, f_pred, '--', label="Predicha (EQL)", linewidth=2, color='C1')
        plt.xlabel("x", fontsize=12)
        plt.ylabel("y", fontsize=12)
        plt.title(f"Nguyen-{func_id} | R²_test = {r2_val:.3f}",
                  fontsize=14, weight='bold')
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()

    # 2D: mapa de contornos
    elif dim == 2:
        n_points_2d = 200
        x1 = np.linspace(x_range[0], x_range[1], n_points_2d)
        x2 = np.linspace(x_range[0], x_range[1], n_points_2d)
        X1, X2 = np.meshgrid(x1, x2)

        if func_id == 9:
            f_true = np.sin(X1) + np.sin(X2**2)
        elif func_id == 10:
            f_true = 2*np.sin(X1)*np.cos(X2)
        elif func_id == 11:
            f_true = X1**X2
        elif func_id == 12:
            f_true = X1**4 - X1**3 + 0.5 * X2**2 - X2

        X_flat = np.stack([X1.flatten(), X2.flatten()], axis=1).astype(np.float32)
        f_pred = model.predict(X_flat, verbose=0).reshape(X1.shape)

        # R² en test si existe
        r2_val = metrics["R2_test"] if metrics else None

        # --- figura ---
        fig, axs = plt.subplots(1, 2, figsize=(10,4))
        cs0 = axs[0].contourf(X1, X2, f_true, levels=50, cmap='viridis')
        cs1 = axs[1].contourf(X1, X2, f_pred, levels=50, cmap='viridis')
        fig.colorbar(cs0, ax=axs[0]); fig.colorbar(cs1, ax=axs[1])
        axs[0].set_title("Función real", weight='bold')
        axs[1].set_title(f"Predicha (EQL)" + (f"\nR²_test = {r2_val:.3f}" if r2_val else ""), weight='bold')
        for ax in axs:
            ax.set_xlabel("x₁"); ax.set_ylabel("x₂")
        plt.tight_layout()
        plt.show()
```

```python
import numpy as np
import pandas as pd
import torch
import physo
import matplotlib.pyplot as plt

# --- 1. Cargar Datos ---
# (Asumo que tu carga de datos funciona, usamos los arrays X_data, y_data)
# file_path = ...
# data = pd.read_csv(...)
# input_features_raw = ["K", "Rem", "prandtl"]
# X_data = data[input_features_raw].values
# y_data = data["nusselt"].values

# --- 2. PRE-PROCESAMIENTO (CRÍTICO) ---

# A. Renombrar variables para evitar el error 'SympifyError: Rem'
# Usaremos 'Re' en lugar de 'Rem'.
X_names = ["K", "Re", "Pr"]  # Nombres seguros para SymPy
y_name  = "Nu"

# B. Escalado de Datos
# PhySO funciona con una red neuronal. Si le das un Reynolds de 50,000, 
# la red se satura y no aprende. Vamos a dividir por valores típicos.
# Esto no cambia la física, solo las unidades internas.

scale_K = 1.0
scale_Re = 1000.0  # Dividimos Reynolds por 1000 (ej: 50000 -> 50)
scale_Pr = 1.0
scale_Nu = 1.0     # Si Nusselt es muy grande (>100), considera dividirlo también

scales_X = np.array([scale_K, scale_Re, scale_Pr])
X_data_scaled = X_data / scales_X
y_data_scaled = y_data / scale_Nu

# Convertir a tensores
X = torch.tensor(X_data_scaled.T, dtype=torch.float32)
y = torch.tensor(y_data_scaled, dtype=torch.float32)

print(f"Datos preparados y escalados.")
print(f"Rango Reynolds original: {X_data[:,1].min()} - {X_data[:,1].max()}")
print(f"Rango Reynolds entrada IA: {X[1].min()} - {X[1].max()}")

# --- 3. CONFIGURACIÓN DEL MODELO ---

# Incluimos 'log' y 'exp' para evitar warnings de los priors por defecto,
# aunque para Nusselt las claves suelen ser mul, div y potencias.
op_names = ["add", "sub", "mul", "div", "inv", "neg", "log", "exp"]

# Aumentamos épocas porque los datos reales son difíciles
EPOCHS = 100 

expression, logs = physo.SR(X, y,
    X_names = X_names,
    y_name  = y_name,
    op_names = op_names,
    run_config = physo.config.config0.config0,
    epochs = EPOCHS,
    # Opcional: Si tienes muchos puntos (>1000), usa un subconjunto para ir más rápido
    # max_n_samples = 1000 
)

# --- 4. RESULTADOS Y DE-ESCALADO ---

if expression is not None:
    best_expr = expression
    print("\n--- ¡Expresión Encontrada! ---")
    
    # La fórmula que te da PhySO está basada en los datos escalados.
    print(f"Fórmula interna (escalada): {best_expr.get_infix_sympy()}")
    
    # Para obtener la fórmula real, debemos recordar que:
    # Re_interno = Re_real / 1000
    # Por tanto, si la fórmula dice: 0.5 * Re, significa: 0.5 * (Re_real / 1000)
    
    # Evaluación (R2)
    y_pred = best_expr(X).detach().numpy()
    from sklearn.metrics import r2_score
    r2 = r2_score(y, y_pred)
    print(f"R^2 Score: {r2:.4f}")

    # --- Gráfico de Paridad (Diagnóstico) ---
    plt.figure(figsize=(6, 6))
    plt.scatter(y.numpy(), y_pred, alpha=0.5)
    plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--')
    plt.xlabel('Nusselt Real (Escalado)')
    plt.ylabel('Nusselt Predicho (Escalado)')
    plt.title(f'Ajuste del Modelo (R2={r2:.2f})')
    plt.show()

else:
    print("No se encontró expresión. Intenta aumentar las épocas.")
```

```python
import matplotlib.pyplot as plt
import numpy as np
import torch

def plot_paper_style(model_expr, X_tensor, y_tensor, scales_X, scale_Nu):
    """
    Genera un gráfico Nu vs Re comparando Datos Reales vs Modelo.
    Maneja el des-escalado para mostrar unidades físicas reales.
    """
    
    # 1. Recuperar datos reales (Des-escalar)
    # X_tensor es [3, N_puntos], asumimos orden: 0:K, 1:Re, 2:Pr
    X_real = X_tensor.numpy().T * scales_X 
    y_real = y_tensor.numpy() * scale_Nu
    
    Re_real = X_real[:, 1] # Columna de Reynolds
    Nu_real = y_real       # Columna de Nusselt
    
    # 2. Generar línea de predicción suave
    # Creamos un vector de Reynolds suave desde el min al max
    re_smooth_real = np.linspace(Re_real.min(), Re_real.max(), 200)
    
    # Para predecir, necesitamos pasarle al modelo los 3 inputs.
    # Como queremos graficar Nu vs Re, fijaremos K y Pr a sus valores promedio
    # Esto es estándar para visualizar correlaciones multivariables en 2D.
    k_mean_real = X_real[:, 0].mean()
    pr_mean_real = X_real[:, 2].mean()
    
    # Preparamos el tensor de entrada para la línea suave (y volvemos a escalar para el modelo)
    # Recordar: El modelo "piensa" en valores escalados.
    re_smooth_scaled = re_smooth_real / scales_X[1]
    k_fixed_scaled   = k_mean_real / scales_X[0]
    pr_fixed_scaled  = pr_mean_real / scales_X[2]
    
    # Construimos la matriz [N, 3] y luego transponemos a [3, N] si es necesario
    # PhySO espera [n_dim, n_samples] usualmente, o checkear tu X.shape
    # En tu código anterior usaste X = torch.tensor(X_data.T), así que espera [Features, Samples]
    
    X_smooth_scaled = np.vstack([
        np.full_like(re_smooth_scaled, k_fixed_scaled), # K fijo
        re_smooth_scaled,                               # Re variando
        np.full_like(re_smooth_scaled, pr_fixed_scaled) # Pr fijo
    ])
    
    X_smooth_tensor = torch.tensor(X_smooth_scaled, dtype=torch.float32)
    
    # 3. Predicción del modelo
    with torch.no_grad():
        y_pred_smooth_scaled = model_expr(X_smooth_tensor).detach().numpy()
    
    # Des-escalar predicción
    y_pred_smooth_real = y_pred_smooth_scaled * scale_Nu
    
    # 4. PLOTEO
    fig, ax = plt.subplots(figsize=(8, 6))
    
    # Datos experimentales (Puntos)
    # Usamos c=Prandtl para mostrar si la dispersión se debe a Pr
    sc = ax.scatter(Re_real, Nu_real, c=X_real[:, 2], cmap='viridis', 
                    alpha=0.6, edgecolors='k', label='Datos Experimentales')
    
    # Modelo (Línea)
    label_model = f"Modelo PhySO\n(con Pr={pr_mean_real:.2f})"
    ax.plot(re_smooth_real, y_pred_smooth_real, 'r-', linewidth=3, label=label_model)
    
    # Formato estilo Paper
    ax.set_xlabel(r'Número de Reynolds ($Re$)', fontsize=14)
    ax.set_ylabel(r'Número de Nusselt ($Nu$)', fontsize=14)
    ax.set_title('Correlación de Transferencia de Calor', fontsize=16)
    
    # Escala logarítmica es MUY común en estos gráficos (Leyes de potencia son rectas en log-log)
    # Descomenta si prefieres ver rectas:
    # ax.set_xscale('log')
    # ax.set_yscale('log')
    
    ax.grid(True, which="both", ls="--", alpha=0.5)
    ax.legend(fontsize=12)
    
    # Barra de color para Prandtl
    cbar = plt.colorbar(sc)
    cbar.set_label(r'Número de Prandtl ($Pr$)', rotation=270, labelpad=15)
    
    plt.tight_layout()
    plt.show()

# --- EJECUTAR DESPUÉS DE OBTENER 'expression' ---
# Asegúrate de pasar las escalas que definimos en el paso anterior
if expression is not None:
    plot_paper_style(expression, X, y, scales_X, scale_Nu)
```
